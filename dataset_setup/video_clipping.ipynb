This notebook houses the code to clip the long videos to 10 second clips.

```python
import os
from pathlib import Path

from moviepy.video.io.ffmpeg_tools import ffmpeg_extract_subclip
from tqdm import tqdm
```

# Extracting 10 seconds from each video

```python
full_length_videos = Path(os.getcwd(), "..\\dataset\\cam")

output_directory = Path(os.getcwd(), "..\\dataset\\video_clips")

video_count = 1
for directory in tqdm(list(full_length_videos.iterdir())):
    for video in directory.iterdir():
        ffmpeg_extract_subclip(video, 0, 10, targetname=Path
        (output_directory, video.name))
    video_count += 1
        
# for i, video_path in enumerate(input_videos):
#     ffmpeg_extract_subclip(video_path, 0, 10, targetname=output_videos[i])
```
