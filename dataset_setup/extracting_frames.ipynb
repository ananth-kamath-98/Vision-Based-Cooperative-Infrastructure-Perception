This notebook houses the code to extract every frame from the video clips 
that make up our dataset. These frames will then be annotated to create 
ground-truths necessary for model evaluation.

```python
import os 
import cv2
from pathlib import Path
from tqdm import tqdm
```

```python
def extract_frames(video_path, output_path):
    if not os.path.exists(output_path):
        os.makedirs(output_path)
    
    vid = cv2.VideoCapture(video_path)
    success, frame = vid.read()
    count = 0
    
    while success:
        frame_path = os.path.join(output_path, f"frame_{count:04d}.jpg")
        cv2.imwrite(frame_path, frame)
        
        success, frame = vid.read()
        count += 1
        
    vid.release()
    
```

```python
dataset_path = Path(os.getcwd(), "..\\dataset")

videos = Path(dataset_path, "video_clips")
frames_output_dir = Path(dataset_path, "frames\\raw")

for video in tqdm(list(videos.iterdir())):
    if not video.is_file():
        break
    
    output_dir = Path(frames_output_dir, video.name.split(".")[0])
    output_dir.mkdir(exist_ok=True)
    
    extract_frames(video, output_dir)
```
